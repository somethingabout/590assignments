#### Add the year the movie was created to the dataframe (another column that has year).


```python
import requests
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup  #this is the spell that helps you read tags"
   
rotten_url = 'https://editorial.rottentomatoes.com/guide/best-movies-21st-century/'
# Use requests to retrieve data from a given URL
rotten_response = requests.get(rotten_url)
# Parse the whole HTML page using BeautifulSoup
rotten_soup = BeautifulSoup(rotten_response.text, 'html.parser')
    
# titles of movies from rotten IMDB
titles = [] # empty list
tags = rotten_soup.find_all('a', {'class':"meta-title"})  #get the tags for the titles

titles=[]
for link in tags: # for each link in the list tags
    x = link.get_text()  # get the text within the node in that list item link
    titles.append(x) #append it to the list called titles
    
scores= []    
tags_scores = rotten_soup.find_all('span', {'class':'tMeterScore'})  #scores from rotten tomatoes\n",
for link,j in zip(tags_scores, range(201)): # zipping a list of the tag scores with 200, this way the first item of tag scores is matched with a number
    if j%2 == 0: # if the item is even, put it in the list
        x = link.get_text()  # get the text inside the node
        scores.append(x) # append the list of the titles"    

titlesdf = pd.DataFrame(titles) # put a list into a dataframe
scoresdf = pd.DataFrame(scores) # put a list into a dataframe      
rottendf = pd.concat([titlesdf, scoresdf], axis = 1)
#create column names
rottendf.columns='titles','scores'
```
